```python
from sklearn.model_selection import KFold
import pandas as pd 
import numpy as np
import random
import os 
```

```python
default_path = os.getcwd()
data_path = os.path.join(default_path, '../data/BWS/score')
base_model = os.path.join(default_path, '../base-model')
config_path = os.path.join(default_path, '../config')
log_path = os.path.join(default_path, '../log')
model_path = os.path.join(default_path, '../model')
config_file = "bert-base.json"
```

```python
criteria = 4
```

```python
data_train = pd.read_csv(os.path.join(data_path, f'bws_a{criteria}_score_train.csv'))
data_val = pd.read_csv(os.path.join(data_path, f'bws_a{criteria}_score_val.csv'))
data_test = pd.read_csv(os.path.join(data_path, f'bws_a{criteria}_score_test.csv'))
```

```python
data_trainset = pd.concat([data_train, data_val])
data_trainset.head(3)
```

```python
data_trainset.reset_index(drop=True, inplace=True)
```

```python
kf = KFold(n_splits=4)
kf.get_n_splits(data_trainset)
```

```python
data_test
```

```python
data_fold_train_dict = dict(); data_fold_test_dict = dict()

for i, (train_index, test_index) in enumerate(kf.split(data_trainset)):
    data_fold_train_dict[f'fold_{i}'] = train_index
    data_fold_test_dict[f'fold_{i}'] = test_index 
    print(f'fold {i}')
    print(f"Train: index={len(train_index)}")
    print(f"Test:  index={len(test_index)}", end='\n\n')
```

```python
data_fold1_train = data_trainset.loc[data_fold_train_dict['fold_0']]; data_fold1_test = data_trainset.loc[data_fold_test_dict['fold_0']]
data_fold2_train = data_trainset.loc[data_fold_train_dict['fold_1']]; data_fold2_test = data_trainset.loc[data_fold_test_dict['fold_1']]
data_fold3_train = data_trainset.loc[data_fold_train_dict['fold_2']]; data_fold3_test = data_trainset.loc[data_fold_test_dict['fold_2']]
data_fold4_train = data_trainset.loc[data_fold_train_dict['fold_3']]; data_fold4_test = data_trainset.loc[data_fold_test_dict['fold_3']] 
```

```python
data_fold1_test
```

```python
data_fold1_train.to_csv(os.path.join(data_path, f'a{criteria}_fold1_train.csv'), index=False)
data_fold1_test.to_csv(os.path.join(data_path, f'a{criteria}_fold1_test.csv'), index=False)
data_fold2_train.to_csv(os.path.join(data_path, f'a{criteria}_fold2_train.csv'), index=False)
data_fold2_test.to_csv(os.path.join(data_path, f'a{criteria}_fold2_test.csv'), index=False)
data_fold3_train.to_csv(os.path.join(data_path, f'a{criteria}_fold3_train.csv'), index=False)
data_fold3_test.to_csv(os.path.join(data_path, f'a{criteria}_fold3_test.csv'), index=False)
data_fold4_train.to_csv(os.path.join(data_path, f'a{criteria}_fold4_train.csv'), index=False)
data_fold4_test.to_csv(os.path.join(data_path, f'a{criteria}_fold4_test.csv'), index=False)
```
