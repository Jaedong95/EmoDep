```python
import pandas as pd
import os 
import nltk
import numpy as np
```

```python
default_path = os.getcwd()
data_path = os.path.join(default_path, '../data')
save_path = os.path.join(data_path, 'BWS')
```

```python
group_a = pd.read_csv(os.path.join(data_path, 'a8_group_a.csv'))
group_b = pd.read_csv(os.path.join(data_path, 'a8_group_b.csv'))
```

```python
group_a.drop_duplicates('text', inplace=True)
group_b.drop_duplicates('text', inplace=True)
```

```python
group_a.to_csv(os.path.join(data_path, 'a8_group_a.csv'), index=False)
group_b.to_csv(os.path.join(data_path, 'a8_group_b.csv'), index=False)
```

```python
a1_sample = group_a.sample(1200, random_state=42) 
daily_sample = group_b.sample(400, random_state=42) 

a1_sample.reset_index(inplace=True, drop=True)
daily_sample.reset_index(inplace=True, drop=True)
```
